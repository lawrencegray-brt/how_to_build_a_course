## Solution: Task 1 - Setup & Imports

````python
# Standard imports for ML regression pipeline
from sklearn.datasets import load_diabetes  # Note: load_diabetes, not fetch!
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np

# For Jupyter notebooks
%matplotlib inline

print("✅ Imports complete!")
````

## Solution: Task 2 - Load & Explore Data

````python
# Load the diabetes dataset
print("Loading Diabetes dataset...")
data = load_diabetes()

# Extract features and target
X = data.data
y = data.target

# Explore the data
print(f"✅ Data loaded successfully!")
print(f"\nDataset Information:")
print(f"  Shape of X: {X.shape}")
print(f"  Shape of y: {y.shape}")
print(f"  Number of patients: {X.shape[0]}")
print(f"  Number of features: {X.shape[1]}")
print(f"\nFeature names: {data.feature_names}")
print(f"\nTarget: Quantitative measure of disease progression one year after baseline")
````

### Expected Output:
- Shape of X: (442, 10)
- Shape of y: (442,)
- 442 patients, 10 features

### Answers to Discussion Questions:
- **How many patients?** 442
- **How many features?** 10
- **Feature names:** age, sex, bmi, bp, s1, s2, s3, s4, s5, s6
- **Do they make sense medically?** Yes! BMI, blood pressure (bp), and blood serum measurements (s1-s6) are all relevant to diabetes.

### Common Student Mistakes:
- Not extracting X and y from the data object
- Confusing data.data with just data

### If students are stuck:
**Hint:** "The structure is the same as California Housing - you have a data object with .data and .target attributes."

---

## Solution: Task 3 - Split Data

````python
# Split data: 80% training, 20% testing
print("Splitting data into training and test sets...")

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% for testing
    random_state=42      # For reproducibility
)

print("✅ Split complete!")
print(f"\nTraining set:")
print(f"  X_train shape: {X_train.shape}")
print(f"  y_train shape: {y_train.shape}")
print(f"\nTest set:")
print(f"  X_test shape: {X_test.shape}")
print(f"  y_test shape: {y_test.shape}")
print(f"\nSplit ratio: {len(X_train)/len(X):.1%} train, {len(X_test)/len(X):.1%} test")
````

### Expected Output:
- X_train shape: (353, 10)
- y_train shape: (353,)
- X_test shape: (89, 10)
- y_test shape: (89,)
- 80/20 split confirmed

### Checkpoint Answer:
Yes! Training ~353, test ~89 ✓

### Teaching Moment:
If students ask "Why is this step so important?", remind them:
> "This prevents data leakage! We split BEFORE any preprocessing. The test set simulates real-world deployment where we encounter new, unseen data."

### Common Student Mistakes:
- Forgetting random_state=42
- Wrong order of return values (mixing up X and y)
- Using 0.8 instead of 0.2 for test_size

### If students are stuck:
**Hint:** "This is identical to what we did in the live demo. Four return values: X_train, X_test, y_train, y_test."

## Solution: Task 4 - Train Model

````python
# Create and train the model
print("Creating Linear Regression model...")
model = LinearRegression()

print("Training model on training data...")
model.fit(X_train, y_train)

print("✅ Model trained successfully!")
print(f"\nModel learned {len(model.coef_)} coefficients (one per feature)")

# Display coefficients with feature names
print(f"\nCoefficients (slopes for each feature):")
print(f"{'Feature':>10s} {'Coefficient':>15s}")
print("-" * 30)
for feature, coef in zip(data.feature_names, model.coef_):
    print(f"{feature:>10s} {coef:15.2f}")

print(f"\nIntercept (baseline): {model.intercept_:.2f}")
````

### Expected Output (Approximate):
Coefficients will vary slightly based on random_state, but typically:
- **bmi**: ~500-600 (large positive - most important!)
- **bp**: ~300-400 (positive)
- **s5**: ~700-800 (large positive)
- **sex**: ~-200 (negative)

### Discussion Question Answers:

**Q: Which features have largest absolute coefficients?**
**A:** Typically bmi, s5, and bp

**Q: What does positive vs negative coefficient mean?**
**A:** 
- **Positive**: Feature increase → disease progression increases (e.g., higher BMI → worse progression)
- **Negative**: Feature increase → disease progression decreases

### Common Student Mistakes:
- Trying to predict before fitting (NotFittedError)
- Not zipping feature names with coefficients

### If students are stuck:
**Hint:** "Remember the sklearn pattern: Create, Fit, Predict. We're on the Fit step."

---

## Solution: Task 5 - Make Predictions

````python
# Make predictions on test set
print("Making predictions on test set...")
y_pred = model.predict(X_test)

print("✅ Predictions complete!")
print(f"\nPredictions shape: {y_pred.shape}")

# Compare first 10 predictions vs actual values
print(f"\nFirst 10 predictions vs actual values:")
print(f"{'Predicted':>12s} {'Actual':>12s} {'Residual':>12s}")
print("-" * 40)
for i in range(10):
    residual = y_test[i] - y_pred[i]
    print(f"{y_pred[i]:12.2f} {y_test[i]:12.2f} {residual:12.2f}")
````

### Expected Observations:
- Predictions are close but not perfect ✓
- Some predictions are too high, some too low ✓
- Residuals (differences) vary in magnitude ✓

### Common Student Mistakes:
- Using X_train instead of X_test for predictions
- Calculating difference as pred - actual (should be actual - pred for residuals)

### If students are stuck:
**Hint:** "Use model.predict() on X_test. Then compare predictions to y_test."

---

## Solution: Task 6 - Evaluate Model

````python
# Calculate evaluation metrics
print("Calculating evaluation metrics...")

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Display results
print("\n" + "=" * 60)
print("MODEL EVALUATION METRICS")
print("=" * 60)

print(f"\nMean Squared Error (MSE):     {mse:.2f}")
print(f"  └─ Average squared error")

print(f"\nRoot Mean Squared Error (RMSE): {rmse:.2f}")
print(f"  └─ Average error in original units")
print(f"  └─ On average, predictions are off by {rmse:.2f} units")

print(f"\nR² Score: {r2:.4f}")
print(f"  └─ Model explains {r2*100:.1f}% of variance")
print(f"  └─ Range: 0 (no predictive power) to 1 (perfect)")

print("\n" + "=" * 60)

# Interpretation
print("\nInterpretation:")
if r2 > 0.6:
    print("✅ Strong model - explains most variance")
elif r2 > 0.4:
    print("✓ Reasonable model - explains fair amount of variance")
    print("  Could be improved with feature engineering or complex models")
else:
    print("⚠️  Weak model - limited predictive power")
````

### Expected Results:
With random_state=42:
- **MSE**: ~2,900 - 3,200
- **RMSE**: ~54 - 57
- **R²**: ~0.45 - 0.52

### Interpretation Guidance:

**My R² score is: ~0.48**

This means my model explains **48%** of variance in disease progression.

**Is this good?**
- For social sciences or medical data: Reasonable
- For physical sciences: Would want higher
- For this educational exercise: Perfect! ✓

### Common Student Mistakes:
- Wrong order of arguments: mean_squared_error(y_pred, y_test) ← Wrong!
- Forgetting np.sqrt() for RMSE
- Evaluating on training data instead of test data

### If students are stuck:
**Hint:** "You need three things: mean_squared_error, np.sqrt, and r2_score. Check the live demo for exact syntax."

## Solution: Task 7 - Residual Plot

````python
# Calculate residuals
residuals = y_test - y_pred

# Create residual plot
plt.figure(figsize=(10, 6))

# Scatter plot
plt.scatter(y_pred, residuals, alpha=0.6, edgecolors='k', linewidth=0.5)

# Reference line at zero
plt.axhline(y=0, color='red', linestyle='--', linewidth=2, 
            label='Zero residuals (perfect predictions)')

# Labels and formatting
plt.xlabel('Predicted Values (disease progression)', fontsize=12)
plt.ylabel('Residuals (Actual - Predicted)', fontsize=12)
plt.title('Residual Plot for Diabetes Dataset', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()

# Analysis
print("\n" + "=" * 60)
print("RESIDUAL PLOT ANALYSIS")
print("=" * 60)
print("\nWhat we observe:")
print("  • Random scatter around zero: ✓ Yes")
print("  • No obvious pattern: ✓ Yes")
print("  • Roughly constant variance: ✓ Mostly")
print("  • Some outliers: ✓ Normal")
print("\nConclusion:")
print("  Linear regression appears appropriate for this dataset.")
print("  No major violations of assumptions detected.")
print("=" * 60)
````

### Expected Residual Plot:
- Random scatter around y=0 ✓
- No clear curve or pattern ✓
- Fairly constant variance (slight increase at higher predictions is acceptable) ✓
- A few outliers (normal!) ✓

### Analysis Question Answers:

**1. Random scatter or pattern?**
Mostly random scatter, which is good!

**2. Funnel shape?**
Slight increase in variance at higher predicted values, but not severe.

**3. Curve?**
No obvious curved pattern.

**4. Is linear regression appropriate?**
Yes! The residual plot shows:
- Random scatter (no systematic bias)
- No nonlinear pattern
- Relatively constant variance

Linear regression is appropriate, though not perfect (R² = 0.48).

### Common Student Mistakes:
- Residuals calculated as pred - actual (should be actual - pred)
- Forgetting plt.show()
- Not adding the zero reference line
- Missing axis labels

### If students are stuck:
**Hint:** "This is identical to the live demo. Scatter plot with y_pred on x-axis, residuals on y-axis. Don't forget the red horizontal line at y=0."

## BONUS Solution: Feature Importance

````python
# Identify top 3 most important features by coefficient magnitude
print("Feature Importance Analysis")
print("=" * 60)

# Create list of (feature, coefficient) tuples
feature_importance = list(zip(data.feature_names, model.coef_))

# Sort by absolute value of coefficient (descending)
feature_importance_sorted = sorted(feature_importance, 
                                   key=lambda x: abs(x[1]), 
                                   reverse=True)

# Display all features ranked
print("\nAll Features Ranked by Importance:")
print(f"{'Rank':>4s} {'Feature':>10s} {'Coefficient':>15s} {'Abs Value':>12s}")
print("-" * 50)
for i, (feature, coef) in enumerate(feature_importance_sorted, 1):
    print(f"{i:4d} {feature:>10s} {coef:15.2f} {abs(coef):12.2f}")

# Top 3
print("\n" + "=" * 60)
print("TOP 3 MOST IMPORTANT FEATURES")
print("=" * 60)
for i in range(3):
    feature, coef = feature_importance_sorted[i]
    direction = "increases" if coef > 0 else "decreases"
    print(f"\n{i+1}. {feature.upper()}")
    print(f"   Coefficient: {coef:.2f}")
    print(f"   Impact: {direction} disease progression")
    print(f"   Magnitude: {abs(coef):.2f} (larger = more important)")

print("\n" + "=" * 60)
````

### Expected Top 3 Features:
With random_state=42, typically:
1. **s5** (blood serum measurement) - ~700-800 (positive)
2. **bmi** (Body Mass Index) - ~500-600 (positive)
3. **bp** (Blood Pressure) - ~300-400 (positive)

### Medical Interpretation:

**The top 3 features affecting disease progression are:**

1. **s5** (blood serum measurement)
2. **bmi** (Body Mass Index)
3. **bp** (Blood Pressure)

**This makes medical sense because:**
- Higher BMI is strongly associated with diabetes complications
- Blood pressure is a key indicator in diabetes management
- Blood serum measurements (like s5, which may represent triglycerides or glucose) directly reflect metabolic health

All three have **positive** coefficients, meaning increases in these values correlate with worse disease progression.

### Teaching Moment:
If students complete this, ask:
> "Does this align with what you know about diabetes? Why would BMI and blood pressure be important?"

This connects ML back to domain knowledge - a critical skill!

### Common Student Mistakes:
- Not using abs() for sorting (would rank by value, not magnitude)
- Forgetting reverse=True (would show least important first)
- Not using lambda function correctly
